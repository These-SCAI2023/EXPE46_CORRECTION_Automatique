```python
from google.colab import drive
from posixpath import split
import glob
import os
import jamspell



def ouverture_fichier(filename):
    print("lecture fichier",filename)
    f = open(filename, "r")
    textlu = f.read()
    f.close()
    print("Lecture OK")
    return textlu

def ecriture_fichier(filename,Texte):
    print("Ecriture du fichier",filename)
    w = open(filename, "w")
    w.write(Texte)
        #w.write('\n')
    w.close()
    print(filename,"Ecriture OK")

def nom_fichier(chemin, moteurcorr,mocr):
  chemin = chemin.split("/")
  output="/".join(chemin[:9])
  output =output+f"/{chemin[7]}_{mocr}-{moteurcorr}/"+chemin[-1].split(".txt")[0]+"-"+moteurcorr+".txt"
  return(output)

print("démarrage----------------")
drive.mount('/content/drive', force_remount=True)
# moteur_ocr=["kraken","tesseract"]
moteur_ocr=["tesseract0.3.10","kraken4.3.13.dev25","lectaurep-kraken4.3.13.dev25"]
moteur_corr=["jspll-ELTeC","jspll-pretrain"]
mcorr=moteur_corr[0]
print("initialiser le correcteur")
jsp = jamspell.TSpellCorrector()   # initialiser le correcteur
print("spécifier le modèle de langue : ", mcorr)
assert jsp.LoadLangModel('efr.bin') # spécifier le modèle de langue
path_ocr=f"/content/drive/MyDrive/Colab Notebooks/EXPE24_JAMSPELL/small-ELTeC-fra2024DF/*/OCR"
# path_test=f"/content/drive/MyDrive/Colab Notebooks/EXPE24_JAMSPELL/TEST_small-TGB-RevueCorpus/*/*OCR"

for ocr in moteur_ocr:
  for path in glob.glob(f"{path_ocr}/*_{ocr}/*.txt"):
    print(path)
    texte = ouverture_fichier(path)
    print(texte[:100])
    name_file= nom_fichier(path, mcorr,ocr)
    print("name_file---> ",name_file)
    ##os.mkdir(name_file)
    if os.path.exists(name_file) == True:
      print("Already DONE : ", name_file)
      continue
    # print("----------------------------------------------->", path)
    # print("----------------------------------------------->", name_file)
    texte_corr=jsp.FixFragment(texte)
    ecriture_fichier(name_file,texte_corr)
```
